## Matematika Diskrit

Matematika diskrit adalah studi tentang struktur matematika yang pada dasarnya diskrit daripada kontinu. Berbeda dengan kalkulus, yang berurusan dengan laju perubahan yang mulus, matematika diskrit berfokus pada objek-objek yang dapat dihitung seperti bilangan bulat, graf, dan pernyataan logika. Relevansinya dalam ilmu komputer sangat besar, karena menjadi dasar bagi desain algoritma, struktur data, kriptografi, dan teori jaringan.

## **Logika**

Logika proposisional adalah dasar dari penalaran dan komputasi. Ini berurusan dengan proposisi (pernyataan yang benar atau salah) dan operasi yang menghubungkannya:

- **AND (**`**dan**`**)**: Benar hanya jika kedua proposisi benar.
- **OR (**`**atau**`**)**: Benar jika setidaknya salah satu proposisi benar.
- **NOT (**`**bukan**`**)**: Membalik nilai kebenaran dari sebuah proposisi.

```python
# Mendefinisikan variabel boolean
p = True
q = False

# Evaluasi ekspresi logika
hasil_and = p and q
hasil_or = p or q
hasil_not_p = not p
ekspresi_kompleks = (p or q) and (not q)

print(f"p = {p}, q = {q}")
print(f"p AND q: {hasil_and}")
print(f"p OR q: {hasil_or}")
print(f"NOT p: {hasil_not_p}")
print(f"(p OR q) AND (NOT q): {ekspresi_kompleks}")
```

**Penjelasan Kode:**

- Kode ini menunjukkan bagaimana operator logika Python (`and`, `or`, `not`) secara langsung mengimplementasikan rekan-rekan logis mereka.
- `p and q` bernilai `False` karena `q` adalah `False`.
- `p or q` bernilai `True` karena `p` adalah `True`.
- Operator boolean bawaan Python ini memungkinkan evaluasi langsung dari ekspresi logika, yang menjadi fondasi dari semua kontrol alur komputasi.

## **Teori Himpunan**

Teori himpunan adalah studi tentang himpunan, yang merupakan kumpulan objek yang berbeda. Operasi dasar pada himpunan meliputi:

- **Gabungan (Union)**: Elemen yang ada di salah satu himpunan atau keduanya.
- **Irisan (Intersection)**: Elemen yang ada di kedua himpunan.
- **Selisih (Difference)**: Elemen yang ada di himpunan pertama tetapi tidak di himpunan kedua.
- **Selisih Simetris (Symmetric Difference)**: Elemen yang ada di salah satu himpunan, tetapi tidak di keduanya.


```python
# Membuat dua himpunan
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Melakukan operasi himpunan
gabungan = A | B       # atau A.union(B)
irisan = A & B        # atau A.intersection(B)
selisih = A - B       # atau A.difference(B)
selisih_simetris = A ^ B # atau A.symmetric_difference(B)

print("Himpunan A:", A)
print("Himpunan B:", B)
print("Gabungan (A | B):", gabungan)
print("Irisan (A & B):", irisan)
print("Selisih (A - B):", selisih)
print("Selisih Simetris (A ^ B):", selisih_simetris)
```

**Penjelasan Kode:**

- `{1, 2, 3, 4}`: Sintaks ini membuat objek `set` bawaan Python.
- Operator `|`, `&`, `-`, dan `^` secara langsung mencerminkan notasi matematika standar untuk operasi himpunan.
- Implementasi `set` yang efisien ini memungkinkan manipulasi kumpulan data yang kompleks dengan cara yang secara teoretis solid dan mudah dibaca.

## **Kombinatorika**

Kombinatorika adalah cabang matematika yang berfokus pada penghitungan. Dua konsep sentral adalah:

- **Permutasi**: Susunan objek di mana urutan penting.
- **Kombinasi**: Seleksi objek di mana urutan tidak penting.

```python
import itertools

items = ['A', 'B', 'C']

# Menghasilkan semua permutasi dari 2 item
permutasi = list(itertools.permutations(items, 2))

# Menghasilkan semua kombinasi dari 2 item
kombinasi = list(itertools.combinations(items, 2))

print("Daftar Item:", items)
print("\nSemua Permutasi 2 item:", permutasi)
print("Jumlah Permutasi:", len(permutasi))
print("\nSemua Kombinasi 2 item:", kombinasi)
print("Jumlah Kombinasi:", len(kombinasi))
```

**Penjelasan Kode:**

- `itertools.permutations(items, 2)`: Menghasilkan semua susunan berurutan dari 2 item yang diambil dari `items`. Perhatikan bahwa `('A', 'B')` dan `('B', 'A')` dianggap berbeda.
- `itertools.combinations(items, 2)`: Menghasilkan semua seleksi tidak berurutan dari 2 item. Di sini, `('A', 'B')` dan `('B', 'A')` dianggap sebagai kombinasi yang sama.
- Modul `itertools` menyediakan alat yang dioptimalkan untuk menghasilkan objek-objek kombinatorial ini, mengubah definisi teoretis menjadi generator yang dapat digunakan secara praktis.

## **Teori Graf**

Teori graf mempelajari graf, yang merupakan struktur matematika yang digunakan untuk memodelkan hubungan antar objek. Representasi umum untuk graf adalah **daftar kedekatan (adjacency list)**, di mana untuk setiap simpul (node), kita menyimpan daftar simpul yang terhubung dengannya.

```python
from collections import deque

# Merepresentasikan graf menggunakan daftar kedekatan (dictionary)
graf = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E'],
    'G': []
}

def find_path_bfs(graph, start, end):
    """Menemukan jalur antara dua node menggunakan Breadth-First Search (BFS)."""
    if start not in graph or end not in graph:
        return None
    queue = deque([(start, [start])])
    visited = {start}
    while queue:
        current_node, path = queue.popleft()
        if current_node == end:
            return path
        for neighbor in graph.get(current_node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                new_path = list(path)
                new_path.append(neighbor)
                queue.append((neighbor, new_path))
    return None

jalur = find_path_bfs(graf, 'A', 'F')
print("Graf (Daftar Kedekatan):", graf)
print("\nJalur dari A ke F menggunakan BFS:", jalur)
```

**Penjelasan Kode:**

- `graf = {...}`: `Dictionary` Python digunakan di mana setiap kunci adalah simpul, dan nilainya adalah daftar tetangganya. Ini adalah implementasi langsung dari daftar kedekatan.
- Fungsi `find_path_bfs` mengimplementasikan algoritma Breadth-First Search, yang menjelajahi graf lapis demi lapis untuk menemukan jalur terpendek dalam graf yang tidak berbobot.
- Dengan cara ini, struktur data Python yang fleksibel memungkinkan representasi yang efisien dan penerapan algoritma teoretis graf.

## **Teori Bilangan**

Teori bilangan adalah studi tentang bilangan bulat dan sifat-sifatnya. Konsep fundamental adalah **pembagi persekutuan terbesar (Greatest Common Divisor - GCD)** dari dua bilangan bulat, yang merupakan bilangan bulat positif terbesar yang membagi keduanya tanpa sisa.

```python
import math

a = 48
b = 18

gcd_val = math.gcd(a, b)

print(f"Pembagi Persekutuan Terbesar (GCD) dari {a} dan {b} adalah: {gcd_val}")
```

**Penjelasan Kode:**

- `math.gcd(a, b)`: Fungsi ini secara efisien mengimplementasikan algoritma Euklides untuk menemukan GCD dari dua bilangan bulat.
- Penggunaan fungsi bawaan ini secara langsung menerapkan algoritma teoretis yang terkenal tanpa perlu implementasi manual, yang menunjukkan kekuatan library standar.

## **Struktur Aljabar**

Struktur aljabar seperti grup mendefinisikan himpunan bersama dengan operasi yang mematuhi aksioma tertentu. **Grup** terdiri dari sebuah himpunan dan satu operasi yang memenuhi sifat penutupan (closure), asosiativitas, elemen identitas, dan elemen invers. Contoh sederhana adalah himpunan bilangan bulat modulo `n` dengan operasi penjumlahan.

```python
# Mensimulasikan penjumlahan dalam grup siklik Z_7 (bilangan bulat modulo 7)
n = 7
a = 5
b = 4

# Operasi "penjumlahan"
hasil = (a + b) % n

print(f"Dalam Z_{n}:")
print(f"{a} + {b} = {hasil} (mod {n})")

# Elemen identitas adalah 0
identitas = 0
hasil_identitas = (a + identitas) % n
print(f"{a} + {identitas} = {hasil_identitas} (mod {n})")
```

**Penjelasan Kode:**

- Operator `%` (modulo) di Python menghitung sisa dari pembagian, yang merupakan alat sempurna untuk mensimulasikan aritmetika dalam grup terbatas.
- `(a + b) % n`: Ini menunjukkan sifat penutupan; hasil penjumlahan dua elemen dalam himpunan `{0, ..., n-1}` akan selalu menjadi elemen lain dalam himpunan yang sama.
- Kode ini mendemonstrasikan bagaimana konsep abstrak grup dapat dimodelkan secara komputasi menggunakan operasi aritmetika sederhana.
