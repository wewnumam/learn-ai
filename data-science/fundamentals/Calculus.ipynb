# Kalkulus

Kalkulus adalah studi tentang perubahan dan merupakan alat yang sangat diperlukan dalam dunia komputasi. Konsep-konsepnya, terutama turunan, menjadi inti dari banyak algoritma modern. Dalam pembelajaran mesin, turunan digunakan dalam algoritma optimisasi seperti _gradient descent_ untuk meminimalkan fungsi kesalahan. Dalam simulasi fisika, mereka memodelkan laju perubahan dalam sistem dinamis. Bagian ini akan mengeksplorasi bagaimana konsep-konsep kalkulus kunci dapat diimplementasikan secara simbolis dalam Python.

## **Turunan dan Turunan Parsial**

**Turunan** suatu fungsi, yang menurut `mitres_18_001_f17_full_book.pdf` dilambangkan sebagai `dy/dx`, mengukur laju perubahan sesaat dari fungsi tersebut. Ini adalah kemiringan garis singgung pada suatu titik di kurva fungsi. Ketika kita berurusan dengan fungsi yang memiliki banyak variabel masukan (fungsi multivariabel), kita menggunakan **turunan parsial**. Turunan parsial dari sebuah fungsi sehubungan dengan satu variabel adalah turunan dari fungsi tersebut, di mana semua variabel lain diperlakukan sebagai konstanta.

Library `SymPy` di Python sangat ideal untuk kalkulus karena memungkinkan komputasi simbolik, artinya ia bekerja dengan ekspresi matematika itu sendiri, bukan hanya nilai numeriknya.

```python
import sympy as sp

# Mendefinisikan variabel simbolik
x, y = sp.symbols('x y')

# 1. Turunan dari fungsi variabel tunggal
f = x**3 + 2*x**2 - 5
turunan_f = sp.diff(f, x)

print("Fungsi f(x):", f)
print("Turunan df/dx:", turunan_f)

# 2. Turunan parsial dari fungsi multivariabel
g = x**2 * y**3 + sp.sin(x)
turunan_parsial_gx = sp.diff(g, x)
turunan_parsial_gy = sp.diff(g, y)

print("\nFungsi g(x, y):", g)
print("Turunan Parsial ∂g/∂x:", turunan_parsial_gx)
print("Turunan Parsial ∂g/∂y:", turunan_parsial_gy)
```

**Penjelasan Kode:**

- `sp.symbols('x y')`: Mendefinisikan `x` dan `y` sebagai variabel simbolik, yang memungkinkan `SymPy` untuk memanipulasinya secara aljabar.
- `f = x**3 + ...`: Membuat ekspresi simbolik untuk fungsi `f(x)`.
- `sp.diff(f, x)`: Menghitung turunan dari `f` terhadap `x`. `SymPy` secara otomatis menerapkan aturan diferensiasi untuk menemukan turunan yang eksak, `3*x**2 + 4*x`.
- `sp.diff(g, x)` dan `sp.diff(g, y)`: Menghitung turunan parsial dari `g` terhadap `x` dan `y` secara terpisah, dengan memperlakukan variabel lainnya sebagai konstanta.
- Implementasi ini secara langsung mencerminkan proses teoretis diferensiasi, menghasilkan turunan analitik yang tepat alih-alih aproksimasi numerik.

## **Aturan Rantai Derivasi**

Aturan Rantai adalah metode fundamental dalam kalkulus untuk menemukan turunan dari fungsi komposit. Jika kita memiliki fungsi `g(x)` yang merupakan komposisi dari dua fungsi lain, katakanlah `g(x) = f(u(x))`, maka turunan `g` adalah produk dari turunan `f` terhadap `u` dan turunan `u` terhadap `x`. Secara matematis: `dg/dx = (df/du) * (du/dx)`.

```python
import sympy as sp

# Mendefinisikan variabel simbolik
x, u = sp.symbols('x u')

# Mendefinisikan dua fungsi
f = sp.sin(u)
u_func = x**2

# 1. Membuat fungsi komposit g(x) = f(u(x))
g = f.subs(u, u_func)
print("Fungsi Komposit g(x):", g)

# 2. Menghitung turunan g(x) secara langsung
turunan_langsung_g = sp.diff(g, x)
print("Turunan dg/dx (Langsung):", turunan_langsung_g)

# 3. Menerapkan Aturan Rantai secara manual
turunan_f_terhadap_u = sp.diff(f, u)
turunan_u_terhadap_x = sp.diff(u_func, x)
turunan_f_di_u_func = turunan_f_terhadap_u.subs(u, u_func)
hasil_aturan_rantai = turunan_f_di_u_func * turunan_u_terhadap_x

print("Hasil Aturan Rantai (df/du * du/dx):", hasil_aturan_rantai)
print("Apakah hasilnya sama:", turunan_langsung_g == hasil_aturan_rantai)
```

**Penjelasan Kode:**

- `g = f.subs(u, u_func)`: Membuat fungsi komposit `g(x)` dengan mengganti `u` dalam `f` dengan ekspresi untuk `u_func`, menghasilkan `sin(x**2)`.
- `sp.diff(g, x)`: Menghitung turunan `sin(x**2)` secara langsung, yang menghasilkan `2*x*cos(x**2)`.
- Bagian ketiga secara eksplisit menguraikan Aturan Rantai: ia menghitung `df/du` dan `du/dx` secara terpisah, lalu mengalikannya.
- Verifikasi `True` di akhir secara komputasi menunjukkan bahwa kedua metode tersebut menghasilkan hasil yang identik, yang mengkonfirmasi validitas Aturan Rantai.

## **Gradien, Jacobian, dan Hessian**

Dalam kalkulus multivariabel, kita sering kali perlu menggeneralisasi konsep turunan ke dimensi yang lebih tinggi.

- **Gradien**: Untuk fungsi bernilai skalar `f(x₁, x₂, ...)` gradien (`∇f`) adalah vektor dari semua turunan parsialnya. Vektor ini menunjuk ke arah kenaikan terbesar dari fungsi.
- **Jacobian**: Untuk fungsi bernilai vektor `F` dari Rⁿ ke Rᵐ, matriks Jacobian `J_F` adalah matriks m x n di mana entri (i, j) adalah turunan parsial dari komponen ke-i fungsi output terhadap variabel input ke-j (∂fᵢ/∂xⱼ). Ini merepresentasikan peregangan, rotasi, dan geseran lokal terbaik yang ditimbulkan oleh fungsi tersebut.
- **Hessian**: Untuk fungsi bernilai skalar, matriks Hessian adalah matriks persegi dari turunan parsial orde kedua. Ini menggambarkan kelengkungan lokal dari fungsi tersebut.

```python
import sympy as sp

# Mendefinisikan variabel simbolik
x, y, z = sp.symbols('x y z')

# 1. Menghitung Gradien
f_scalar = x**2 * y
variabel_f = sp.Matrix([x, y])
gradien_f = sp.Matrix([sp.diff(f_scalar, var) for var in variabel_f])

print("Fungsi Skalar:", f_scalar)
print("Vektor Gradien ∇f:\n", gradien_f)

# 2. Menghitung Matriks Jacobian
F_vektor = sp.Matrix([x*y, sp.sin(x), y**2])
variabel_F = sp.Matrix([x, y])
jacobian_F = F_vektor.jacobian(variabel_F)

print("\nFungsi Vektor F:", F_vektor)
print("Matriks Jacobian J_F:\n", jacobian_F)

# 3. Menghitung Matriks Hessian
hessian_f = sp.hessian(f_scalar, (x, y))

print("\nFungsi Skalar:", f_scalar)
print("Matriks Hessian H_f:\n", hessian_f)
```

**Penjelasan Kode:**

- **Gradien**: Kode ini menghitung turunan parsial dari `f_scalar` terhadap `x` dan `y`, lalu menyusunnya ke dalam sebuah matriks kolom (vektor). Output `[2*x*y, x**2]` adalah gradien dari `f`.
- **Jacobian**: `F_vektor.jacobian(variabel_F)` secara otomatis menghitung matriks Jacobian. Baris pertama `[y, x]` adalah turunan parsial dari `x*y` terhadap `x` dan `y`. Baris kedua `[cos(x), 0]` adalah turunan parsial dari `sin(x)`.
- **Hessian**: `sp.hessian(f_scalar, (x, y))` menghitung matriks dari turunan parsial orde kedua, yang menangkap kelengkungan fungsi.
- Penggunaan `SymPy` memungkinkan komputasi otomatis dari struktur turunan tingkat tinggi ini, menerjemahkan definisi kalkulus multivariabel menjadi operasi yang dapat dieksekusi.
