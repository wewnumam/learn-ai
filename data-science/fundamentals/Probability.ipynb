# Peluang

Peluang, atau probabilitas, adalah kerangka kerja matematika untuk mengukur dan menganalisis ketidakpastian. Ini adalah fondasi dari statistik dan ilmu data. Seperti yang dinyatakan dalam `IntroductoryStatistics-OP_i6tAI7e.pdf`, probabilitas adalah "alat matematika yang digunakan untuk mempelajari keacakan." Bagian ini mencakup konsep-konsep dasar yang menjadi dasar bagi model statistik.

## **Dasar-Dasar Peluang**

Teori peluang dibangun di atas konsep inti seperti **ruang sampel** (semua hasil yang mungkin) dan **peristiwa**. Salah satu aturan fundamental yang dibahas dalam `IntroductoryStatistics-OP_i6tAI7e.pdf` adalah **aturan komplemen**: `P(A') = 1 - P(A)`. Selain itu, **Hukum Bilangan Besar (Law of Large Numbers)** menyatakan bahwa seiring dengan meningkatnya jumlah percobaan, probabilitas empiris (yang diamati) dari suatu peristiwa akan mendekati probabilitas teoretisnya.

```python
import numpy as np

def simulasi_lempar_dadu(jumlah_lemparan):
    """Mensimulasikan pelemparan dadu dan menghitung probabilitas empiris."""
    lemparan = np.random.randint(1, 7, size=jumlah_lemparan)
    jumlah_sukses = np.sum(lemparan > 4)
    return jumlah_sukses / jumlah_lemparan

prob_teoretis = 2/6
lemparan_1000000 = simulasi_lempar_dadu(1000000)

print(f"Probabilitas Teoretis mendapatkan angka > 4: {prob_teoretis:.5f}")
print(f"Probabilitas Empiris (1,000,000 lemparan): {lemparan_1000000:.5f}")
```

**Penjelasan Kode:**

- `np.random.randint(...)`: Menghasilkan bilangan bulat acak untuk mensimulasikan pelemparan dadu.
- Kode ini menghitung proporsi lemparan yang hasilnya lebih besar dari 4.
- Seperti yang ditunjukkan oleh output, saat jumlah lemparan meningkat, probabilitas empiris yang dihitung dari simulasi semakin mendekati probabilitas teoretis.
- Ini adalah demonstrasi komputasi yang kuat dari Hukum Bilangan Besar, yang menghubungkan hasil percobaan dunia nyata dengan konsep probabilitas teoretis.

## **Teorema Bayes**

Teorema Bayes adalah hasil fundamental dalam teori peluang yang menggambarkan cara memperbarui keyakinan kita tentang suatu hipotesis berdasarkan bukti baru. Ini didasarkan pada konsep **probabilitas bersyarat** (`P(A|B)`), yaitu probabilitas peristiwa `A` terjadi, _mengingat_ bahwa peristiwa `B` telah terjadi, sebagaimana dijelaskan dalam `IntroductoryStatistics-OP_i6tAI7e.pdf`.

`P(A|B) = [P(B|A) * P(A)] / P(B)`

- `P(A|B)`: Probabilitas posterior (keyakinan yang diperbarui).
- `P(A)`: Probabilitas prior (keyakinan awal).
- `P(B|A)`: Likelihood (kemungkinan bukti, mengingat hipotesis).

```python
def teorema_bayes(p_prior, p_likelihood, p_bukti_given_not_hipotesis):
    """Menghitung probabilitas posterior menggunakan Teorema Bayes."""
    p_not_hipotesis = 1 - p_prior
    p_bukti = (p_likelihood * p_prior) + (p_bukti_given_not_hipotesis * p_not_hipotesis)
    p_posterior = (p_likelihood * p_prior) / p_bukti
    return p_posterior

# Parameter: probabilitas memiliki penyakit mengingat tes positif
prior = 0.01      # P(Penyakit) = 1% prevalensi
likelihood = 0.99   # P(Positif|Penyakit) = 99% sensitivitas
p_fp = 0.05       # P(Positif|Sehat) = 5% tingkat positif palsu

posterior = teorema_bayes(prior, likelihood, p_fp)

print(f"Probabilitas Prior (memiliki penyakit): {prior:.2%}")
print(f"Probabilitas Posterior (setelah tes positif): {posterior:.2%}")
```

**Penjelasan Kode:**

- Fungsi ini secara langsung mengimplementasikan rumus Teorema Bayes untuk masalah diagnostik medis klasik.
- **Prior**: Keyakinan awal kita adalah hanya 1% dari populasi yang memiliki penyakit.
- **Posterior**: Output menunjukkan bahwa bahkan dengan tes positif, probabilitas Anda benar-benar memiliki penyakit hanya sekitar 16,6%.
- Kode ini menerjemahkan rumus teoretis Bayes menjadi alat komputasi yang menunjukkan bagaimana keyakinan diperbarui secara rasional dengan adanya bukti baru.

## **Variabel Acak dan Fungsi Kepadatan Peluang (PDF)**

**Variabel Acak (Random Variable - RV)** adalah variabel yang nilainya merupakan hasil numerik dari fenomena acak. `IntroductoryStatistics-OP_i6tAI7e.pdf` membedakan antara RV **diskrit** (nilai yang dapat dihitung, mis., jumlah buku) dan **kontinu** (nilai dalam rentang, mis., panjang panggilan telepon). Untuk RV kontinu, kita menggunakan **Fungsi Kepadatan Peluang (PDF)**, `f(x)`. Probabilitas dihitung sebagai area di bawah kurva PDF.

```python
from scipy.stats import norm

# Bekerja dengan Distribusi Normal (mean=0, std_dev=1)
mean = 0
std_dev = 1

# 1. Menghitung nilai PDF pada titik tertentu (tinggi kurva, bukan probabilitas)
pdf_value = norm.pdf(0, loc=mean, scale=std_dev)
print(f"Nilai PDF pada x=0: {pdf_value:.4f}")

# 2. Menghitung probabilitas kumulatif (CDF), P(X <= x)
cdf_value_at_0 = norm.cdf(0, loc=mean, scale=std_dev)
print(f"Probabilitas Kumulatif P(X <= 0): {cdf_value_at_0:.4f}")

# Menghitung P(-1 < X < 1) = CDF(1) - CDF(-1)
prob_range = norm.cdf(1, loc=mean, scale=std_dev) - norm.cdf(-1, loc=mean, scale=std_dev)
print(f"Probabilitas P(-1 < X < 1): {prob_range:.4f} (sekitar 68%)")
```

**Penjelasan Kode:**

- `norm.pdf(x, ...)`: Menghitung nilai PDF dari distribusi normal pada titik `x`.
- `norm.cdf(x, ...)`: Menghitung probabilitas kumulatif hingga `x`, yang mewakili area di bawah kurva PDF dari -∞ hingga `x`.
- Kode ini menunjukkan bagaimana menghitung probabilitas untuk variabel kontinu sebagai area, secara langsung menerapkan konsep teoretis PDF dan CDF untuk menghasilkan probabilitas numerik.

## **Jenis-Jenis Distribusi**

Berbagai jenis fenomena acak dimodelkan oleh distribusi probabilitas yang berbeda. Beberapa yang umum meliputi:

- **Distribusi Binomial**: Menurut `IntroductoryStatistics-OP_i6tAI7e.pdf`, distribusi ini muncul dari "percobaan Bernoulli" di mana ada sejumlah percobaan `n` yang tetap dengan dua hasil. Ini memodelkan jumlah keberhasilan dalam `n` percobaan.
- **Distribusi Normal**: Distribusi berbentuk lonceng yang ada di mana-mana.
- **Distribusi Eksponensial**: Sering digunakan untuk memodelkan waktu antara peristiwa.

```python
from scipy.stats import binom, expon
import numpy as np

# 1. Menghasilkan angka acak dari Distribusi Binomial
# Skenario: Lempar koin 10 kali (n=10) dengan p=0.7, ulangi 5 kali.
hasil_binomial = binom.rvs(n=10, p=0.7, size=5)
print(f"Hasil Binomial (jumlah keberhasilan dalam 10 percobaan):")
print(hasil_binomial)

# 2. Menghasilkan angka acak dari Distribusi Eksponensial
# Skenario: Waktu tunggu dengan rata-rata 5 menit.
hasil_eksponensial = expon.rvs(scale=5, size=5)
print(f"\nHasil Eksponensial (waktu tunggu yang disimulasikan):")
print(np.round(hasil_eksponensial, 2))
```

**Penjelasan Kode:**

- `binom.rvs(...)`: Menghasilkan variabel acak (`r`andom `v`ariate`s`) dari distribusi binomial. Outputnya adalah array yang menunjukkan jumlah keberhasilan untuk setiap eksperimen.
- `expon.rvs(...)`: Menghasilkan variabel acak dari distribusi eksponensial, mensimulasikan waktu tunggu.
- Kode ini menunjukkan bagaimana fungsi library dapat digunakan untuk menghasilkan data sampel yang mengikuti distribusi teoretis tertentu, yang memungkinkan simulasi dan pemodelan skenario dunia nyata.
