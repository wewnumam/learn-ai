# Statistik

Statistik adalah ilmu mengumpulkan, menganalisis, menginterpretasikan, dan menyajikan data. Dalam dunia yang semakin didorong oleh data, statistik memberikan metodologi untuk membuat keputusan yang terinformasi. Fondasi dari statistik adalah teori peluang, yang menurut `IntroductoryStatistics-OP_i6tAI7e.pdf`, adalah "alat matematika yang digunakan untuk mempelajari keacakan."

## **Konsep Statistik Dasar**

Untuk memahami sebuah set data, kita sering kali memulai dengan menghitung beberapa metrik ringkasan. Berdasarkan `IntroductoryStatistics-OP_i6tAI7e.pdf`, ini biasanya dibagi menjadi dua kategori:

- **Ukuran Tendensi Sentral**: Metrik ini mendeskripsikan "pusat" dari data, seperti **Mean** (rata-rata), **Median** (nilai tengah), dan **Mode** (nilai yang paling sering muncul).
- **Ukuran Variabilitas**: Metrik ini mendeskripsikan "sebaran" data, seperti **Varians** dan **Standar Deviasi**.

```python
import numpy as np
from scipy import stats

data = np.array([1, 2, 2, 3, 4, 5, 5, 5, 6, 7])

# Menghitung ukuran tendensi sentral
mean_val = np.mean(data)
median_val = np.median(data)
mode_val = stats.mode(data)

# Menghitung ukuran variabilitas
variance_val = np.var(data)
std_dev_val = np.std(data)

print("Set Data:", data)
print(f"\nMean: {mean_val:.2f}")
print(f"Median: {median_val}")
print(f"Mode: Nilai {mode_val.mode} muncul {mode_val.count} kali")
print(f"\nVarians: {variance_val:.2f}")
print(f"Standar Deviasi: {std_dev_val:.2f}")
```

**Penjelasan Kode:**

- `np.mean()`, `np.median()`, `stats.mode()`, `np.var()`, dan `np.std()`: Fungsi-fungsi ini dari NumPy dan SciPy secara langsung menghitung metrik statistik dasar.
- Output memberikan ringkasan kuantitatif tentang distribusi data: pusat data berada di sekitar 4, dengan nilai yang paling umum adalah 5, dan data tersebar rata-rata sekitar 2.05 unit dari mean.
- Ini menunjukkan bagaimana fungsi library menerjemahkan definisi statistik teoretis menjadi nilai-nilai konkret yang meringkas data.

## **Statistik Deskriptif**

Statistik deskriptif bertujuan untuk meringkas dan mengatur data. Selain ukuran di atas, **kuartil** dan **persentil** sangat berguna. Seperti yang dinyatakan dalam `IntroductoryStatistics-OP_i6tAI7e.pdf`, "persentil menunjukkan kedudukan relatif dari suatu nilai data." Kuartil membagi data menjadi empat bagian yang sama: Kuartil Pertama (Q1), Median (Q2), dan Kuartil Ketiga (Q3). Ringkasan lima angka (minimum, Q1, median, Q3, maksimum) memberikan gambaran menyeluruh tentang distribusi.

```python
import numpy as np
import pandas as pd

data = np.array([65, 78, 82, 85, 88, 90, 91, 92, 95, 98, 100])

# Menggunakan NumPy untuk menghitung persentil
q1 = np.percentile(data, 25)
median = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Set Data:", data)
print(f"Kuartil Pertama (Q1 / Persentil ke-25): {q1}")
print(f"Median (Q2 / Persentil ke-50): {median}")
print(f"Kuartil Ketiga (Q3 / Persentil ke-75): {q3}")

# Menggunakan Pandas untuk ringkasan statistik yang komprehensif
df = pd.DataFrame(data, columns=['Nilai'])
print("\nRingkasan Statistik Deskriptif dari Pandas:\n", df['Nilai'].describe())
```

**Penjelasan Kode:**

- `np.percentile(data, 25)`: Menghitung nilai di mana 25% data berada di bawahnya, sesuai dengan definisi Q1.
- Metode `.describe()` dari Pandas dengan cepat menghasilkan ringkasan statistik deskriptif utama, termasuk ringkasan lima angka (`min`, `25%`, `50%`, `75%`, `max`).
- Kode ini mendemonstrasikan bagaimana alat komputasi dapat secara efisien menghasilkan ringkasan statistik yang mendalam yang sejalan dengan prinsip-prinsip deskriptif teoretis.

## **Grafik & Bagan**

Visualisasi data sangat penting untuk memahami distribusi. Dua jenis plot yang umum adalah:

- **Histogram**: Seperti yang dijelaskan dalam `IntroductoryStatistics-OP_i6tAI7e.pdf`, "sebuah histogram... terdiri dari batang-batang dengan lebar yang sama yang digambar berdampingan." Tinggi setiap batang merepresentasikan frekuensi data dalam interval tertentu.
- **Box Plot**: Memberikan representasi visual dari ringkasan lima angka, yang sangat baik untuk mengidentifikasi kemiringan dan pencilan.

```python
import numpy as np
import matplotlib.pyplot as plt

# Menghasilkan data sampel dari distribusi normal
np.random.seed(42)
data = np.random.normal(loc=100, scale=15, size=200)

# 1. Membuat Histogram
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.hist(data, bins=15, edgecolor='black')
plt.title('Histogram Distribusi Data')
plt.xlabel('Nilai')
plt.ylabel('Frekuensi')

# 2. Membuat Box Plot
plt.subplot(1, 2, 2)
plt.boxplot(data, vert=False)
plt.title('Box Plot Distribusi Data')
plt.xlabel('Nilai')

print("Kode untuk menghasilkan histogram dan box plot telah disiapkan.")
print("Jika dijalankan di lingkungan dengan GUI, plt.show() akan menampilkan visualisasi.")
```

**Penjelasan Kode:**

- `plt.hist(data, ...)` dan `plt.boxplot(data, ...)`: Fungsi-fungsi Matplotlib ini mengambil data mentah dan menghasilkan visualisasi standar yang didasarkan pada prinsip-prinsip statistik.
- **Interpretasi Visual**: Histogram memungkinkan kita untuk melihat bentuk distribusi data, sementara box plot secara visual menyajikan ringkasan lima angka, menghubungkan metrik numerik dari bagian sebelumnya ke representasi grafis.
- Plot-plot ini adalah implementasi langsung dari alat-alat teoretis yang digunakan untuk mengeksplorasi data, yang menjembatani analisis kuantitatif dan pemahaman intuitif.

## **Statistik Inferensial**

Statistik inferensial memungkinkan kita untuk membuat kesimpulan tentang **populasi** berdasarkan data dari **sampel**. Alat utamanya adalah **uji hipotesis**. Proses ini, sebagaimana dijelaskan dalam `IntroductoryStatistics-OP_i6tAI7e.pdf`, melibatkan perumusan dua hipotesis yang berlawanan:

- **Hipotesis Nol (H₀)**: Pernyataan tentang tidak adanya efek atau tidak adanya perbedaan.
- **Hipotesis Alternatif (Hₐ)**: Pernyataan yang kita yakini benar jika kita menolak hipotesis nol.

**Uji-t satu sampel** digunakan untuk menentukan apakah rata-rata populasi berbeda dari nilai spesifik.

```python
from scipy import stats
import numpy as np

# Misalkan kita memiliki sampel waktu respon (dalam detik)
sampel_data = np.array([2.5, 2.8, 3.1, 2.9, 3.4, 2.7, 3.0, 3.2])
nilai_hipotesis = 3.0

# H₀: Rata-rata populasi (μ) adalah 3.0 detik.
# Hₐ: Rata-rata populasi (μ) tidak sama dengan 3.0 detik.

t_statistic, p_value = stats.ttest_1samp(sampel_data, popmean=nilai_hipotesis)

print(f"Statistik-t: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretasi hasil
alpha = 0.05
if p_value < alpha:
    print(f"\nKesimpulan: Karena p-value ({p_value:.4f}) < {alpha}, kita menolak hipotesis nol.")
else:
    print(f"\nKesimpulan: Karena p-value ({p_value:.4f}) >= {alpha}, kita gagal menolak hipotesis nol.")
```

**Penjelasan Kode:**

- `stats.ttest_1samp(...)`: Fungsi SciPy ini melakukan semua perhitungan untuk uji-t satu sampel, membandingkan rata-rata sampel dengan nilai hipotesis.
- **Statistik-t**: Mengukur seberapa jauh rata-rata sampel dari nilai hipotesis nol, dalam satuan kesalahan standar.
- **P-value**: Probabilitas mengamati hasil yang sama ekstremnya (atau lebih ekstrem) dengan hasil sampel kita, dengan asumsi hipotesis nol benar. P-value yang kecil (< 0.05) memberikan bukti untuk menolak H₀.
- Kode ini mengotomatiskan kerangka kerja uji hipotesis, mengubah data sampel menjadi kesimpulan statistik yang didasarkan pada teori probabilitas.
