# 1. Identifikasi Kebutuhan Data (Titanic Dataset)

Proyek ini ditujukan untuk mengembangkan model prediktif yang menjawab pertanyaan: “Orang seperti apa yang lebih mungkin selamat dalam insiden tenggelamnya kapal Titanic?” menggunakan data penumpang (misalnya nama, usia, jenis kelamin, kelas sosial ekonomi, dll). Fitur yang relevan antara lain:

| Fitur              | Alasan Penting                                                             | Ketersediaan Data                   |
| ------------------ | -------------------------------------------------------------------------- | ----------------------------------- |
| `Survived`         | Target: 1 = selamat, 0 = meninggal                                         | Tersedia                            |
| `Pclass`           | Kelas tiket → kelas sosial bisa memengaruhi peluang selamat                | Tersedia                            |
| `Sex`              | Jenis kelamin → penumpang wanita lebih mungkin selamat                     | Tersedia                            |
| `Age`              | Usia → anak-anak lebih mungkin diselamatkan                                | Tersedia                            |
| `SibSp`            | Jumlah saudara/ pasangan di kapal → memengaruhi kelangsungan hidup         | Tersedia                            |
| `Parch`            | Jumlah orang tua/anak di kapal → bisa memengaruhi keselamatan              | Tersedia                            |
| `Fare`             | Harga tiket → indikasi status sosial                                       | Tersedia                            |
| `Embarked`         | Pelabuhan naik → beberapa analisis menunjukkan distribusi survival berbeda | Tersedia                            |
| `Cabin` (opsional) | Lokasi kabin → bisa memengaruhi akses ke lifeboat                          | Tersedia tapi banyak missing values |


## 2. Pengambilan Data Menggunakan Python

Pengambilan data menggunakan kagglehub, sehingga bisa langsung digunakan di skrip berikut:

```python
!pip install kagglehub
```

```python
import kagglehub
import pandas as pd
import os # Import os module for path manipulation

# Download dataset Titanic
# kagglehub.dataset_download returns the path to the extracted dataset files
path = kagglehub.dataset_download("yasserh/titanic-dataset")

print("Path dataset:", path)

# List contents of the downloaded directory to verify its structure
print("Contents of downloaded directory:", os.listdir(path))

# Load the single Titanic dataset file
# Based on the contents, it's 'Titanic-Dataset.csv' not separate train/test files.
titanic_data = pd.read_csv(os.path.join(path, "Titanic-Dataset.csv"))

# Preview data
titanic_data.head()
```

# 3. Integrasi Data ke Database atau Repository

 Integrasi data Titanic dilakukan ke dalam database SQLite (titanic_data.db). Ini melibatkan pembuatan koneksi database, pengunggahan pandas DataFrame ke tabel bernama titanic_data, dan verifikasi integrasi dengan melakukan kueri beberapa baris pertama dari database.

```python
import sqlite3

# Buat koneksi database
conn = sqlite3.connect('titanic_data.db')

# Upload data train dan test
titanic_data.to_sql('titanic_data', conn, if_exists='replace', index=False)

# Verifikasi
pd.read_sql('SELECT * FROM titanic_data LIMIT 5', conn)
```
