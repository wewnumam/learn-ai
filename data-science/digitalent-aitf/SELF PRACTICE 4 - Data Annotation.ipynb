```python
import kagglehub
import pandas as pd
import os

pd.set_option('display.max_columns', None)

# Download dataset
path = kagglehub.dataset_download("yasserh/wine-quality-dataset")

print("Path dataset:", path)

# List contents of the downloaded directory to verify its structure
print("Contents of downloaded directory:", os.listdir(path))

# Load the single dataset file
df = pd.read_csv(os.path.join(path, "WineQT.csv"))

display(df.info())
```

# **Data Annotation**

```python
quality_annotation = {
    1: "Sangat Buruk",
    2: "Buruk Sekali",
    3: "Buruk",
    4: "Kurang Baik",
    5: "Cukup",
    6: "Cukup Baik",
    7: "Baik",
    8: "Sangat Baik",
    9: "Istimewa",
    10: "Luar Biasa"
}

df["quality_annotation"] = df["quality"].map(quality_annotation)

df.head()
```
