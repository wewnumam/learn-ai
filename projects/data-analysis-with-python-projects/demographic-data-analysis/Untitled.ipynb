```python
import pandas as pd
```

```python
# Read data from file
df = pd.read_csv('./adult.data.csv')
df.info()
```

---
How many of each race are represented in this dataset? This should be a Pandas series with race names as the index labels.

```python
df['race'].value_counts()
```

---
What is the percentage of people who have a Bachelor's degree?

```python
round((df['education'] == 'Bachelors').mean().item() * 100, 1)
```

---
What percentage of people with advanced education (`Bachelors`, `Masters`, or `Doctorate`) make more than 50K?

```python
higher_education_df = df[df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])]
round((higher_education_df[higher_education_df['salary'] == '>50K'].shape[0] / higher_education_df.shape[0]) * 100, 1)
```

What percentage of people without advanced education make more than 50K?

```python
lower_education_df = df[~df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])]
round((lower_education_df[lower_education_df['salary'] == '>50K'].shape[0] / lower_education_df.shape[0]) * 100, 1)
```

---
What percentage of the people who work the minimum number of hours per week have a salary of >50K?

```python
min_work_hours = df['hours-per-week'].min()
min_workers_df = df[df['hours-per-week'] == min_work_hours]
num_min_workers = min_workers_df.shape[0]    
rich_min_workers = min_workers_df[min_workers_df['salary'] == '>50K'].shape[0]
round((rich_min_workers / num_min_workers) * 100, 1) if num_min_workers > 0 else 0
```

---
What country has the highest percentage of people that earn >50K?

```python
country_counts = df['native-country'].value_counts()
rich_country_counts = df[df['salary'] == '>50K']['native-country'].value_counts()
country_rich_percentage = (rich_country_counts / country_counts * 100).round(1)
print(country_rich_percentage.idxmax(), country_rich_percentage.max())
```

---
Identify the most popular occupation for those who earn >50K in India.

```python
df[(df['salary'] == '>50K') & (df['native-country'] == 'India')]['occupation'].value_counts().idxmax()
```
