```python
(df['education'] == 'Bachelors').mean().item() * 100
```

```python
df['salary'].value_counts()
```

```python
higher_education = df[((df['education'] == 'Bachelors') | 
                            (df['education'] == 'Masters') | 
                            (df['education'] == 'Doctorate')) &
                            (df['salary'] == '>50K')]

(len(higher_education) / len(df)) * 100
```

```python
lower_education = df[((df['education'] != 'Bachelors') & 
                        (df['education'] != 'Masters') & 
                        (df['education'] != 'Doctorate')) &
                        (df['salary'] == '>50K')]

(len(lower_education) / len(df)) * 100
```

```python
min_work_hours = df['hours-per-week'].min().item()
min_work_hours
```

```python
len(df[(df['hours-per-week'] == min_work_hours) & (df['salary'] == '>50K')]) / len(df) * 100
```

```python
richest_country = df[(df['salary'] == '>50K')]['native-country'].value_counts().idxmax()
richest_country
```

```python
len(df[(df['salary'] == '>50K') & (df['native-country'] == richest_country)]) / len(df) * 100
```
