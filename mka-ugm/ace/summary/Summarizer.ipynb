```python
from transformers import pipeline

# Pakai model ringan
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

def rangkum_dari_file(namafile, min_length=20, max_length=60):
    # Baca isi file txt
    with open(namafile, "r", encoding="utf-8") as f:
        paragraf = f.read()

    # Ringkas teks
    summary = summarizer(
        paragraf,
        min_length=min_length,
        max_length=max_length,
        do_sample=False
    )
    return summary[0]['summary_text']


# --- CONTOH PENGGUNAAN ---
if __name__ == "__main__":
    hasil = rangkum_dari_file("paragraf.txt", min_length=30, max_length=60)
    print("Ringkasan:")
    print(hasil)
```
